# Discretization of Continuous Numerical Data

* Numerical data is divided into distinct intervals.
* Through discretization, numerical features can be converted into categorical data.
* `Discretization` refers to the process of transforming a continuous variable into a variable with two or more categories.

```python
import numpy as np
import pandas as pd
```

# Using Numpy

### `np.digitize()`
A method for dividing numerical features based on multiple threshold values.
* The input value for the `bins` parameter represents the left boundary of each interval.
* If you set `bins = [20, 35, 65]`, the data is divided into 4 intervals: `[~ 20)`, `[20, 35)`, `[35, 65)`, `[65 ~)`.
* You can include the right bound to be inclusive, by setting `right = True`


```python
# given data
age = [[6], [12], [20], [39], [66]]  # 2D
age
```

```python
np.digitize(age, bins=[20, 35, 65])
```

```python
np.digitize(age, bins=[20, 30, 65], right=True)
```

# Using sklearn

### `Binarizer()`

Use to convert continuous variables into a binary variable with two values based on a specified threshold:

* If the value is equal to or less than the threshold, it is converted to '0'.
* If the value is greater than the threshold, it is converted to '1'.


```python
from sklearn.preprocessing import Binarizer
```

```python
binarizer = Binarizer(threshold=20)
binarizer.fit_transform(age)
```

## `KBinsDiscretizer()`

This function divides continuous feature values into multiple intervals, and you can specify the number of intervals to divide into.

* `encode` :
    * 'onehot-dense' returns a dense array.
    * 'ordinal' returns sequential categorical values.
* `strategy`:
    * 'quantile': Ensures that each interval contains approximately the same number of data points.
    * 'uniform': Ensures that each interval has the same width.
    * 'kmeans': Values in each bin have the same nearest center of a 1D
* The values of the intervals can be checked using the `bin_edges_` attribute.


```python
from sklearn.preprocessing import KBinsDiscretizer
```

### strategy=uniform

```python
age = np.array([[0], [1], [12], [20], [39], [45], [80]])

kb = KBinsDiscretizer(4, encode="ordinal", strategy="uniform")
kb.fit_transform(age)  # 1~19, 20~39, 40~59, 60~80
```

```python
age = np.array([[0], [1], [12], [20], [39], [45], [80]])

kb = KBinsDiscretizer(4, encode="onehot-dense", strategy="uniform")
kb.fit_transform(age)  # 1~19, 20~39, 40~59, 60~80
```

### Strategy=quantile

```python
age = np.array([[0], [1], [12], [20], [39], [45], [80]])

kb = KBinsDiscretizer(5, encode="ordinal", strategy="quantile")
kb.fit_transform(age)
```

```python
age = np.array([[0], [1], [12], [20], [39], [45], [80]])

kb = KBinsDiscretizer(5, encode="onehot-dense", strategy="quantile")
kb.fit_transform(age)
```

```python
kb.bin_edges_
```

# Using Pandas

Binning allows converting numerical data into categorical data by categorizing the numeric values.

* `pd.cut()`: Divides the data into **uniform intervals width**, or by specifying the boundary values for the intervals.

* `pd.qcut()`: Divides the data into a specified number of intervals, ensuring each interval **contains the equal number of data points**, without explicitly specifying the boundary values.


## pd.cut() - uniform intervals width

Equal-Length Buckets Categorization

* The `pd.cut()` function allows easy categorization by taking numerical data and the interval boundaries as arguments.
* The data segmented using `pd.cut()` is returned as a Series with a categorical data type.


```python
age = np.array([0, 1, 12, 20, 39, 45, 80])  # pd.cut, pd.qcut requires 1D
age
```

```python
cats1 = pd.cut(age, 4)
print(type(cats1))
cats1
```

```python
df = pd.DataFrame({"value": age, "pd.cut": cats1})
df
```

### specifying boundaries possible

The function will automatically divide the data into equal intervals based on the minimum and maximum values of the data. However, you can also specify desired boundaries

The `age` data is returned as a categorical Series with 4 specified intervals


```python
age = np.array([0, 1, 12, 20, 39, 45, 80])
bins = [18, 25, 35, 60, 65]

# 18 ~ 25 / 25 ~ 35 / 35 ~ 60 / 60 ~ 65 total 4 intervals
cats2 = pd.cut(age, bins)
cats2
```

```python
df["pd.cuts.withbin"] = cats2
df
```

### `.codes`
By using `.codes`, you can see the integer index representing which interval each element of `age` belongs to. 

For example, for categorizing case with specified interval, 20-year-old belongs to the first interval (index 0), and 39-year-old belongs to the second interval, and 1-year-old is not belonging anywhere.


```python
print(cats1.codes)
print(cats2.codes)
```

By using `.value_counts()`, you can check the count of elements in each interval. This helps to determine how many elements belong to each category in a categorical Series.

```python
print(cats1.value_counts())  # un-even value counts
print(cats2.value_counts())  # un-even value counts
```

### labeling group names
When calling `pd.cut()`, you can add the `labels = [list]` argument to specify custom category labels for each interval.

```python
age = np.array([0, 1, 12, 20, 39, 45, 80])

# 18 ~ 25 / 25 ~ 35 / 35 ~ 60 / 60 ~ 65 total 4 intervals
cats3 = pd.cut(age, bins=[18, 25, 35, 60, 65], labels=["A", "S", "D", "F"])
cats3
```

```python
df["labels"] = cats3
df
```

## pd.qcut() - equal-Size Buckets Categorization

Pandas provides a function called `qcut()`.

* It defines intervals based on the specified number of intervals (bins).
* Unlike `pd.cut()`, which divides the data based on the minimum and maximum values, 
* `pd.qcut()` takes the data distribution into account and divides the data into intervals such that each interval contains an equal number of data points, using quantiles as the bin boundaries.


```python
age = np.array([0, 1, 12, 20, 39, 45, 80])
cats4 = pd.qcut(age, 4)
cats4
```

* Instead of simply dividing the range between the minimum and maximum values into four equal parts, it considers the distribution and splits the data into quartiles.
* Unlike the `cut()` function, it cannot be said that each interval has the same length.


```python
cats4.value_counts()
```

```python
df["pd.qcut"] = cats4
df
```

```python
df["qcut-label"] = pd.qcut(age, 4, labels=["q", "w", "e", "r"])
df
```

# 🍷 Wine Example

### 1. Discretize

**Objective**: Convert continuous data into categorical data. For example, discretize the `alcohol` or `flavanoids` feature in the wine dataset into different categories.

**Example**: Discretize the `alcohol` feature in the wine dataset into 3 categories.

```python
from sklearn.datasets import load_wine
wine = load_wine()
df = pd.DataFrame(wine.data, columns=wine.feature_names)
df
```

```python
cats = pd.cut(
    df["alcohol"],
    bins=[11, 12, 13, 15, 19],
    labels=["Low", "Medium", "High", "VeryHigh"],
)
print(cats.value_counts())
df["alcohol_discretized"] = cats
df
```
