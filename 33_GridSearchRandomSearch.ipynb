# Model Selection
Setting appropriate hyperparameters is crucial for selecting the best model.

We will cover efficient techniques to choose the optimal model from multiple candidate models.

- Exhaustive Search (Grid Search)
- Random Search

# Iris Data set

```python
from sklearn.datasets import load_iris

iris = load_iris()
X = iris.data
y = iris.target
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(solver="saga", max_iter=5000)
```

# Exhaustive Search (Grid Search)
We use Scikit-learn’s `GridSearchCV`. It trains models on all combinations of hyperparameters and finds the best set of hyperparameters.

```python
from sklearn.model_selection import GridSearchCV

# Define candidate hyperparameter values
penalty = ["l1", "l2"]  # 2 Penalty types
C = [0.01, 0.1, 1, 10, 100]  # 5 Regularization strength

# Create hyperparameter dictionary
hyperparameters = dict(C=C, penalty=penalty)

# GridSearch trains models on all hyperparameter combinations
# The best performing model is selected as the final model
gridsearch = GridSearchCV(model, hyperparameters, cv=5, verbose=0)

# Perform grid search
best_model1 = gridsearch.fit(X, y)
```

`GridSearchCV` uses K-fold cross-validation during training.

- verbose=0 means no output; values 1–3 increase verbosity with more details.
- n_jobs = -1 uses all CPU cores for parallel processing.
- Since all combinations are trained, it can take a lot of time.

In this example:
- 2 penalty options: L1, L2
- 5 values for C
- 5 folds in cross-validation → Total of 2 × 5 × 5 = 50 models are trained and evaluated.

After GridSearchCV finishes, you can check the best hyperparameters.

```python
# Check best hyperparameters
print("Best penalty:", best_model1.best_estimator_.get_params()["penalty"])
print("Best C value:", best_model1.best_estimator_.get_params()["C"])

# Check the performance of the model with the best hyperparameters
print("Best model score:", best_model1.score(X, y))
```

### Interpretation

**Best penalty: l1**  
→ This means that the model achieved the best performance using the **L1 regularization** (Lasso).  
L1 regularization helps make the model simpler and more interpretable by driving unnecessary feature coefficients to zero, effectively performing feature selection.

**Best C value: 10**  
→ Among the values for the regularization strength parameter C=10 was optimal.  
A larger `C` means weaker regularization, while a smaller `C` means stronger regularization.  
Here, a relatively weak regularization yielded the best model performance.

**Best model score: 0.98**  
→ The model trained on the entire dataset with the optimal hyperparameters achieved an accuracy of 98%.  
This indicates that the model learned the data patterns very well.

### Additional interpretation tips

- Since L1 regularization was selected, the model likely ignored (set to zero) some features, effectively removing irrelevant information.  
- The choice of `C=10` suggests that slightly relaxing the regularization helped the model represent the data better without excessive constraints.  
- Although 98% accuracy is high, interpret this score in the context of the data distribution and problem difficulty. Always consider cross-validation scores and test set performance for a comprehensive evaluation.

# Random Search Method
We use Scikit-learn's `RandomizedSearchCV`. Compared to grid search, it reduces the computational cost of selecting the best model by investigating random combinations of hyperparameters.

Set candidate values for hyperparameters, such as the penalty type and the regularization parameter C.

In RandomizedSearchCV, when specifying distributions, hyperparameter values are randomly sampled from these distributions without replacement.


```python
from sklearn.model_selection import RandomizedSearchCV

# Define candidate hyperparameter values
penalty = ["l1", "l2"]  # 2 Penalty types
C = [0.01, 0.1, 1, 10, 100]  # 5 Regularization strength

# Create hyperparameter dictionary
hyperparameters = dict(C=C, penalty=penalty)

# Create a RandomizedSearchCV object
randomizedsearch = RandomizedSearchCV(
    model, hyperparameters, random_state=1, n_iter=100, cv=5, verbose=0, n_jobs=-1
)

best_model2 = randomizedsearch.fit(X, y)
```

```python
from scipy.stats import uniform

# Randomly sample from a uniform distribution between 0 and 100
C = uniform(loc=0, scale=100)
```

```python
# Check the Best Hyperparameters

print("Best penalty:", best_model2.best_estimator_.get_params()["penalty"])
print("Best C value:", best_model2.best_estimator_.get_params()["C"])

# Evaluate the model performance with the optimal hyperparameters
print("Best model score:", best_model2.score(X, y))
```
