## grid-like format
### `plt.subplot`

```python
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.plot(x, 0.5*x)
plt.ylim([-100, 100])


plt.subplot(1, 2, 2)
plt.plot(x, 10*x)
plt.ylim([-100, 100])
```

```python
plt.figure(figsize=(5, 5))

plt.subplot(2, 2, 1)  # rows, columns, panel selected
plt.plot(x, x**2)

plt.subplot(2, 2, 4)
plt.plot(x, 2*x)
```

```python
plt.figure(figsize=(12, 6))

plt.subplot(2, 3, 1)  # rows, columns, panel selected
plt.plot(x, x**2)

plt.subplot(1, 3, 3)
plt.plot(x, x**3)

plt.subplot(2, 3, 5)
plt.plot(x, -5 * x)
```

```python
plt.figure(figsize=(12, 6))
plt.title("My Nice Plot")

plt.subplot(1, 2, 1)  # rows, columns, panel selected
plt.plot(x, x**2)
plt.plot([0, 0, 0], [-10, 0, 100])
plt.legend(["X^2", "Vertical Line"])
plt.xlabel("X")
plt.ylabel("X Squared")

plt.subplot(1, 2, 2)
plt.plot(x, -1 * (x**2))
plt.plot([-10, 0, 10], [-50, -50, -50])
plt.legend(["-X^2", "Horizontal Line"])

plt.xlabel("X")
plt.ylabel("X Squared")

plt.tight_layout()
```

# subplots (axes plotting)

When we call the `subplots()` function we get a tuple containing a `Figure` and a `axes` element.

`axes` has its own plotting method like `df.plot()` from pandas

```python
plot_objects = plt.subplots(figsize=(3,3))
fig, ax = plot_objects
ax.plot([1, 2, 3], [11, 12, 13], ls="dashed")
plot_objects
```

```python
fig, ax = plt.subplots()
ax.plot(x, 5*x)
```

```python
fig1, axes1 = plt.subplots(figsize=(12, 6))  # it is not plt.plot(figsize=(12,6))

axes1.plot(x, 10*x,"m-", label="10X")
axes1.plot(x, (x**2), "c--", label="X^2")
axes1.set_title("My Nice Plot")
axes1.legend()

plt.show()
```

### nrows and ncols

```python
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(nrows=2, ncols=2, figsize=(14, 6))

ax4.plot(np.random.randn(50), c="yellow")
ax1.plot(np.random.randn(50), c="red", ls="--")
ax2.plot(np.random.randn(50), "g:")
ax3.plot(np.random.randn(50), c="blue", marker="o", linewidth=3.0)
```

# Data Visualizing

### Scatter Plot

```python
N = 50
x = np.random.rand(N)
y = np.random.rand(N)
r = 20 * np.random.rand(N)
colors = np.random.rand(N)

area = np.pi * r**2  # 0 to 15 point radii
```

```python
plt.figure(figsize=(14, 6))
plt.scatter(x, y, s=area, c=colors, alpha=0.5, cmap="Spectral")
plt.colorbar()
plt.show()
```

```python
plt.plot(x)
```

```python
x = np.random.rand(50)
y = np.random.rand(50)

plt.scatter(x, y)
```

```python
fig = plt.figure(figsize=(14, 6))
ax1 = fig.add_subplot(1, 2, 1)
plt.scatter(x, y, s=area, c=colors, alpha=0.5, cmap="Pastel1")
plt.colorbar()
ax2 = fig.add_subplot(1, 2, 2)
plt.scatter(x, y, s=area, c=colors, alpha=0.5, cmap="Pastel2")
plt.colorbar()
plt.show()
```

### Histograms

```python
randx = np.random.randn(1000)
randx
```

```python
plt.hist(
    randx,
    bins=15,
    color="yellow",
    edgecolor="red",
)
plt.xlim(xmin=-5, xmax=5)
plt.xlabel("random x")
plt.ylabel("frequency")
plt.show()
```

### KDE (kernel density estimation)

```python
from scipy import stats
density = stats.kde.gaussian_kde(randx)
density
```

```python
plt.plot(figsize=(12, 6))

values2 = np.linspace(min(randx) - 10, max(randx) + 10, 100)
plt.plot(values2, density(values2), color="#FF7F00")
plt.fill_between(values2, 0, density(values2), alpha=0.5, color="#FF7F00")
plt.xlim(xmin=-5, xmax=5)
plt.show()
```

#### Combine plots

```python
plt.subplots(figsize=(12, 6))

plt.hist(
    randx,
    bins=100,
    alpha=0.8,
    density=1,
    histtype="bar",
    color="steelblue",
    edgecolor="green",
)

plt.plot(values2, density(values2), color="#FF7F00", linewidth=3.0)
plt.xlim(xmin=-5, xmax=5)

plt.show()
```

## Bar plots

```python
Y = np.random.rand(1, 5)[0]
Y2 = np.random.rand(1, 5)[0]
print(Y)
print(Y2)
```

```python
plt.figure(figsize=(12, 4))
plt.bar(np.arange(5), Y, width=0.5, color="#00ebbb")
plt.title("my bar plot")
plt.show()
```

Also can be stacked bars, and add a legend to the plot:

```python
plt.bar(
    np.arange(5), Y, color="#bbb894", label="Consumption of Coffee"
)
plt.bar(
    np.arange(5), Y2, color="#e17055", bottom=Y, label="Consumption of Orange Juice",
)

plt.xlabel("Month")
plt.legend()
plt.show()
```
